# Analisis Model 1D MT untuk Menemukan Model Optimal

Notebook ini melakukan analisis dua tahap:
1.  Membaca file log (`.logfile`) untuk menemukan iterasi inversi yang menghasilkan model paling optimal...
2.  Setelah iterasi terbaik ditemukan, notebook akan membaca file data resistivitas (`.iter`) yang sesuai...
3.  HARUS DILETAKKAN PADA DIRECTORY BERISI HASIL INVERSI PyGMI, DIRECTORY PER TITIK INVERSI 1D
4.  Dari sini anda dapat menggunakan outputnya untuk menampilkan pseudomap horizontal dan vertikal (lihat skrip lain)
5.  Pastikan model, iter, resp dan logfile ada semua

# 1. Import Packages & Initialization

```python
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

base_dir = r"C:\KULIAH\KULTER 2\MT\trial-2\edis\AMT" 
input_dir = os.path.join(base_dir, "00. DATA EDI")
output_dir = os.path.join(base_dir, "02. PLOT L-CURVE & 1D MODEL")
os.makedirs(output_dir, exist_ok=True)
```

# 2. Core Parser Functions

```python
def parse_te_logfile(filepath):
    if not os.path.exists(filepath): return [], [], []
    with open(filepath, 'r', errors='ignore') as f: text = f.read()
    iters = [int(m) for m in re.findall(r'\*\*\s*Iteration\s+(\d+)\s*\*\*', text, re.IGNORECASE)]
    rms = [float(m) for m in re.findall(r'and is\s*=\s*([0-9.eE+-]+)', text, re.IGNORECASE)]
    rough = [float(m) for m in re.findall(r'Roughness\s*is\s*=\s*([0-9.eE+-]+)', text, re.IGNORECASE)]
    n = min(len(iters), len(rms), len(rough))
    return iters[:n], rms[:n], rough[:n]

def read_iter(filepath):
    if not os.path.exists(filepath): return None
    with open(filepath, 'r', errors='ignore') as f: text = f.read()
    pc_match = re.search(r'PARAM COUNT:\s*(\d+)', text, re.IGNORECASE)
    pc = int(pc_match.group(1)) if pc_match else None
    res_vals = [float(v) for line in text.splitlines() if not line.strip().startswith('!') and ':' not in line for v in line.split()]
    return 10 ** np.array(res_vals[:pc] if pc else res_vals)

def read_model1d(filepath):
    if not os.path.exists(filepath): return None, None
    with open(filepath, 'r', errors='ignore') as f: lines = f.readlines()
    layers = int(next((line.split()[-1] for line in lines if '#LAYERS:' in line.upper()), 1)) - 1
    depths = []
    started = False
    for line in lines:
        if '! top_depth' in line:
            started = True
            continue
        if started and line.strip() and not line.startswith('!'):
            try: depths.append(float(line.split()[0].replace(',', '.')))
            except ValueError: pass
    return layers, np.array(depths)

def find_lcurve_corner(rough, rms):
    """
    Cari titik elbow (siku) L-curve berdasarkan kurvatura maksimum,
    bukan sekadar RMS minimum. Bekerja di ruang log-log karena
    L-curve klasik memang dianalisis di skala log.

    REVISI:
    - Sumbu x (roughness) dan y (RMS) dinormalisasi ke rentang [0,1]
      sebelum menghitung turunan. Tanpa ini, rumus kurvatura sangat
      sensitif terhadap skala masing-masing sumbu -- pada bagian kurva
      yang sudah landai/flat (RMS sudah konvergen), dx & dy mendekati
      nol sehingga sedikit noise numerik di ddx/ddy bisa "meledak"
      menjadi nilai kurvatura semu yang lebih tinggi dari elbow asli.
      Ini yang menyebabkan beberapa stasiun (mis. MT1, MT2, MT4, MT5,
      MT8, MT9) titik optimalnya jatuh di ekor kurva, bukan di elbow
      visualnya.
    - Window smoothing diperbesar (size=5) untuk meredam noise di
      region flat, terutama karena data cenderung padat (20 iterasi).
    """
    x = np.log10(np.array(rough) + 1e-12)
    y = np.log10(np.array(rms) + 1e-12)

    # normalisasi ke [0,1] agar kedua sumbu punya skala yang sebanding
    x_range = x.max() - x.min()
    y_range = y.max() - y.min()
    x_n = (x - x.min()) / (x_range + 1e-12)
    y_n = (y - y.min()) / (y_range + 1e-12)

    # smoothing biar turunan tidak berisik akibat noise numerik
    if len(x_n) >= 5:
        from scipy.ndimage import uniform_filter1d
        x_s = uniform_filter1d(x_n, size=5)
        y_s = uniform_filter1d(y_n, size=5)
    else:
        x_s, y_s = x_n, y_n

    dx  = np.gradient(x_s)
    dy  = np.gradient(y_s)
    ddx = np.gradient(dx)
    ddy = np.gradient(dy)

    # rumus kurvatura kappa untuk kurva parametrik
    curvature = np.abs(dx * ddy - dy * ddx) / (dx**2 + dy**2 + 1e-12) ** 1.5

    # abaikan 1-2 titik pertama/terakhir (ujung sering "meledak" karena gradient numerik)
    edge = 1 if len(curvature) > 4 else 0
    search_range = curvature[edge: len(curvature) - edge] if edge else curvature
    best_local = np.argmax(search_range)
    best_idx = best_local + edge

    return best_idx
```

# 3. Plotting Functions

```python
def plot_lcurve(iters, rms, rough, best_i, best_r, best_ro, site, out_dir):
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, ax = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle(f"1D Occam Inversion L-Curve Analysis - Station: {site}", fontsize=16, fontweight='bold', color='#1f2b38', y=1.05)
    ax[0].plot(iters, rms, marker='o', color='#d62728', lw=2, ms=7)
    ax[0].axvline(best_i, color='gray', ls='--', alpha=0.7)
    ax[0].set(title="RMS Error vs Iteration", xlabel="Iteration Number", ylabel="RMS Error")
    ax[0].annotate(f'Best RMS: {best_r:.2f}\nIter: {best_i}', xy=(best_i, best_r), xytext=(best_i + max(1, len(iters)*0.05), best_r + (max(rms)-best_r)*0.15), arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=7), fontsize=10, bbox=dict(boxstyle="round,pad=0.4", fc="#f8f9fa", ec="gray", alpha=0.9))
    ax[1].plot(iters, rough, marker='s', color='#1f77b4', lw=2, ms=7)
    ax[1].axvline(best_i, color='gray', ls='--', alpha=0.7)
    ax[1].set(title="Roughness vs Iteration", xlabel="Iteration Number", ylabel="Roughness")
    ax[2].plot(rough, rms, marker='^', color='#2ca02c', lw=2, ms=7)
    ax[2].scatter(best_ro, best_r, color='gold', s=200, zorder=5, ec='black', lw=1.5, label='Optimal Model')
    ax[2].set(title="RMS vs Roughness", xlabel="Roughness", ylabel="RMS Error")
    ax[2].legend(fontsize=11)
    plt.tight_layout()
    plt.savefig(os.path.join(out_dir, f"{site}_L-CURVE.png"), dpi=300, bbox_inches='tight')
    plt.close()

def plot_1d(depths, res, site, iter_num, out_dir):
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, ax = plt.subplots(figsize=(6, 8))
    ax.step(res, depths, where='post', color='#b22222', lw=2.5)
    ax.set(xscale='log', xlabel='Resistivity ($\\Omega \\cdot m$)', ylabel='Depth (m)', title=f'1D Resistivity Model - Station : {site} Iteration {iter_num:03d}')
    ax.invert_yaxis()
    ax.grid(True, which="both", ls='--', alpha=0.6)
    plt.tight_layout()
    plt.savefig(os.path.join(out_dir, f"{site}_1D MODEL.png"), dpi=300, bbox_inches='tight')
    plt.close()
```

# 4. Main Batch Execution Loop

```python
summary = []
if os.path.exists(input_dir):
    for item in os.listdir(input_dir):
        site_folder = os.path.join(input_dir, item)
        if not os.path.isdir(site_folder): continue
        site = item.split('_')[0]
        iters, rms, rough = parse_te_logfile(os.path.join(site_folder, "TE.logfile"))
        if not iters: continue
        best_idx = find_lcurve_corner(rough, rms)
        best_i, best_r, best_ro = iters[best_idx], rms[best_idx], rough[best_idx]
        try:
            plot_lcurve(iters, rms, rough, best_i, best_r, best_ro, site, output_dir)
            summary.append({"Station": site, "Optimal_Iter": best_i, "Optimal_RMS": round(best_r, 4)})
            iter_file = next((os.path.join(site_folder, f) for f in os.listdir(site_folder) if f.lower().endswith('.iter') and re.search(r'\d+', f) and int(re.search(r'\d+', f).group()) == best_i), None)
            res_data = read_iter(iter_file) if iter_file else None
            num_layers, ground_depths = read_model1d(os.path.join(site_folder, "Model1D"))
            if res_data is not None and ground_depths is not None:
                f_depths, f_res = ground_depths[:num_layers], res_data[:num_layers]
                if len(f_res) < num_layers:
                    f_res = np.concatenate((f_res, np.full(num_layers - len(f_res), f_res[-1] if len(f_res)>0 else 0)))
                pd.DataFrame({'Depth': f_depths, 'Resistivity': f_res, 'RMS Error': best_r}).to_csv(os.path.join(output_dir, f"{site}.csv"), index=False)
                plot_1d(f_depths, f_res, site, best_i, output_dir)
        except Exception as e: print(f"{site} Error: {e}")
if summary: print(pd.DataFrame(summary).to_string(index=False))
```
